# EDA and Results Analysis

Multimodal Deep Learning for UK Wholesale Electricity Price Forecasting  
Looking at the data, exploring patterns, and breaking down model results.

## 1. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 5)
plt.rcParams['figure.dpi'] = 120

PROCESSED = os.path.join('..', 'data', 'processed')
```

## 2. Main Dataset Overview

The model_features.csv is what the models actually trained on. ~96k half-hourly rows from Jan 2020 to Jun 2025.

```python
df = pd.read_csv(os.path.join(PROCESSED, 'model_features.csv'), parse_dates=['datetime_local'])
print(f'Rows: {df.shape[0]:,}, Columns: {df.shape[1]}')
print(f'Date range: {df["datetime_local"].min()} to {df["datetime_local"].max()}')
print(f'Memory: {df.memory_usage(deep=True).sum() / 1024**2:.1f} MB')
df.head(3)
```

```python
# Quick null check - I remember the pipeline used interpolation so shouldn't be many
nulls = df.isnull().sum()
nulls[nulls > 0]
```

## 3. Target Variable: System Sell Price

This is what we're trying to forecast. Wholesale electricity price in £/MWh.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

# Full time series
axes[0, 0].plot(df['datetime_local'], df['systemSellPrice'], linewidth=0.3, color='steelblue')
axes[0, 0].set_title('System Sell Price - Full Series (2020-2025)')
axes[0, 0].set_ylabel('£/MWh')

# Distribution
axes[0, 1].hist(df['systemSellPrice'].clip(0, 300), bins=80, color='steelblue', edgecolor='white', alpha=0.7)
axes[0, 1].axvline(df['systemSellPrice'].median(), color='red', ls='--', label=f"Median: {df['systemSellPrice'].median():.1f}")
axes[0, 1].set_title('Price Distribution (clipped at £300)')
axes[0, 1].set_xlabel('£/MWh')
axes[0, 1].legend()

# Weekly profile
weekly = df.groupby('day_of_week')['systemSellPrice'].agg(['mean', 'std'])
axes[1, 0].bar(weekly.index, weekly['mean'], yerr=weekly['std'], capsize=3, color='steelblue', alpha=0.7)
axes[1, 0].set_title('Average Price by Day of Week')
axes[1, 0].set_xticks(range(7))
axes[1, 0].set_xticklabels(['Mon','Tue','Wed','Thu','Fri','Sat','Sun'])
axes[1, 0].set_ylabel('£/MWh')

# Hourly profile
hourly = df.groupby('hour')['systemSellPrice'].agg(['mean', 'std'])
axes[1, 1].plot(hourly.index, hourly['mean'], marker='o', color='steelblue')
axes[1, 1].fill_between(hourly.index, hourly['mean'] - hourly['std'], hourly['mean'] + hourly['std'], alpha=0.2, color='steelblue')
axes[1, 1].set_title('Average Price by Hour of Day')
axes[1, 1].set_xlabel('Hour')
axes[1, 1].set_ylabel('£/MWh')

plt.tight_layout()
plt.show()
```

Observations so far:
- The 2022 spike is very visible - prices went absolutely mental during the energy crisis
- Distribution is right-skewed with a long tail (those crisis prices)
- Weekdays are higher than weekends - makes sense, industrial demand drops off
- The hourly pattern shows the classic morning peak and evening peak (duck curve-ish)

## 4. Crisis Period vs Normal Period

The whole point of this project was to see if sentiment features help specifically during the 2022 crisis.

```python
# Define periods
crisis_start = '2022-02-01'
crisis_end = '2022-10-31'

df['period'] = 'normal'
df.loc[(df['datetime_local'] >= crisis_start) & (df['datetime_local'] <= crisis_end), 'period'] = 'crisis'

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(df['datetime_local'], df['systemSellPrice'], linewidth=0.3, color='lightgray')
crisis_mask = df['period'] == 'crisis'
ax.scatter(df.loc[crisis_mask, 'datetime_local'], df.loc[crisis_mask, 'systemSellPrice'], 
           s=1, color='crimson', alpha=0.3, label='Crisis Period (Feb-Oct 2022)')
ax.axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')
ax.set_title('System Sell Price - Crisis Period Highlighted')
ax.set_ylabel('£/MWh')
ax.legend()
plt.show()

print(f"Normal period mean: {df.loc[df['period']=='normal', 'systemSellPrice'].mean():.1f} £/MWh")
print(f"Crisis period mean: {df.loc[df['period']=='crisis', 'systemSellPrice'].mean():.1f} £/MWh")
print(f"Crisis max: {df.loc[df['period']=='crisis', 'systemSellPrice'].max():.1f} £/MWh")
```

Yeah so crisis prices were literally 2-3x normal. No wonder ARIMA completely falls apart - it can't adapt to regime shifts like this.

## 5. Feature Exploration

Let's look at some key features: generation mix, gas prices, weather, and the NLP sentiment signal.

```python
# Generation mix over time - main fuel types
gen_cols = ['CCGT', 'COAL', 'NUCLEAR', 'WIND', 'BIOMASS', 'OCGT']
gen_names = ['CCGT (Gas)', 'Coal', 'Nuclear', 'Wind', 'Biomass', 'OCGT (Gas Peaker)']

# Downsample to daily for readability
df_daily = df.resample('D', on='datetime_local')[gen_cols].mean()

fig, ax = plt.subplots(figsize=(14, 6))
for col, name in zip(gen_cols, gen_names):
    ax.plot(df_daily.index, df_daily[col], label=name, linewidth=0.8)
ax.axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')
ax.set_title('Daily Average Generation by Fuel Type')
ax.set_ylabel('MW')
ax.legend(ncol=2)
plt.show()
```

```python
# Gas prices vs electricity prices
fig, ax1 = plt.subplots(figsize=(14, 5))

ax1.plot(df['datetime_local'], df['gas_price_close'], color='orange', linewidth=0.5, alpha=0.7, label='TTF Gas Price')
ax1.set_ylabel('Gas Price (€/MWh)', color='orange')
ax1.tick_params(axis='y', labelcolor='orange')

ax2 = ax1.twinx()
ax2.plot(df['datetime_local'], df['systemSellPrice'], color='steelblue', linewidth=0.3, alpha=0.5, label='Electricity Price')
ax2.set_ylabel('Electricity Price (£/MWh)', color='steelblue')
ax2.tick_params(axis='y', labelcolor='steelblue')

ax1.axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')
ax1.set_title('Gas vs Electricity Prices (TTF benchmark)')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')
plt.show()

print(f"Correlation between gas and electricity: {df['gas_price_close'].corr(df['systemSellPrice']):.3f}")
print(f"Crisis-period correlation: {df.loc[crisis_mask, 'gas_price_close'].corr(df.loc[crisis_mask, 'systemSellPrice']):.3f}")
```

Gas and electricity are highly correlated (as expected - gas sets the marginal price in the UK market).
During the crisis the correlation actually drops though - other factors (nuclear outages, low wind, interconnector issues) were all compounding.

```python
# Sentiment score over time
fig, axes = plt.subplots(2, 1, figsize=(14, 8))

# Daily sentiment
sent_daily = df.groupby(df['datetime_local'].dt.date)['sentiment_score'].mean()
axes[0].plot(pd.DatetimeIndex(sent_daily.index), sent_daily.values, linewidth=0.6, color='green')
axes[0].axhline(0, color='gray', ls='--', linewidth=0.5)
axes[0].axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')
axes[0].set_title('Daily News Sentiment Score (FinBERT)')
axes[0].set_ylabel('Sentiment (-1 to +1)')

# Article count over time
art_daily = df.groupby(df['datetime_local'].dt.date)['article_count'].sum()
axes[1].bar(pd.DatetimeIndex(art_daily.index), art_daily.values, width=1, color='steelblue', alpha=0.6)
axes[1].axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')
axes[1].set_title('Daily Article Volume (Guardian API)')
axes[1].set_ylabel('Article count')

plt.tight_layout()
plt.show()
```

Interesting - sentiment dips negative during the crisis period (makes sense, all the news was about energy crisis, Ukraine war etc).
Article volume also spikes during the crisis - more coverage = more uncertainty.

```python
# Weather features
fig, axes = plt.subplots(3, 1, figsize=(14, 9), sharex=True)

axes[0].plot(df['datetime_local'], df['temperature_2m'], linewidth=0.3, color='orange')
axes[0].set_ylabel('Temperature (°C)')
axes[0].axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')

axes[1].plot(df['datetime_local'], df['wind_speed_10m'], linewidth=0.3, color='teal')
axes[1].set_ylabel('Wind Speed (km/h)')
axes[1].axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')

axes[2].plot(df['datetime_local'], df['solar_generation_mw'], linewidth=0.3, color='goldenrod')
axes[2].set_ylabel('Solar Gen (MW)')
axes[2].axvspan(pd.Timestamp(crisis_start), pd.Timestamp(crisis_end), alpha=0.08, color='red')

axes[2].set_xlabel('Date')
fig.suptitle('Weather Features', fontsize=13)
plt.tight_layout()
plt.show()
```

## 6. Model Comparison

Let's look at both the crisis holdout and walk-forward results.

```python
crisis_results = pd.read_csv(os.path.join(PROCESSED, 'model_comparison_crisis.csv'))
wf_results = pd.read_csv(os.path.join(PROCESSED, 'model_comparison_walkforward.csv'))

print('=== CRISIS HOLDOUT ===')
print(crisis_results[['Model', 'Config', 'MAE', 'RMSE', 'sMAPE', 'R2']].to_string(index=False))
print()
print('=== WALK-FORWARD (Avg across 5 folds) ===')
wf_avg = wf_results.groupby(['Model', 'Config'])[['MAE', 'RMSE', 'sMAPE', 'R2']].mean().round(2)
print(wf_avg.to_string())
```

```python
# Visual comparison - crisis holdout
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

crisis_pivot = crisis_results.pivot_table(index='Model', columns='Config', values='MAE', aggfunc='mean').round(2)
crisis_pivot_sorted = crisis_pivot.mean(axis=1).sort_values()

x = np.arange(len(crisis_pivot_sorted))
w = 0.35

for i, model in enumerate(crisis_pivot_sorted.index):
    with_sent = crisis_pivot.loc[model, 'With Sentiment'] if 'With Sentiment' in crisis_pivot.columns else np.nan
    without_sent = crisis_pivot.loc[model, 'Without Sentiment'] if 'Without Sentiment' in crisis_pivot.columns else np.nan
    if i == 0:
        axes[0].bar(i - w/2, without_sent, w, label='Without Sentiment', color='lightcoral', alpha=0.8)
        axes[0].bar(i + w/2, with_sent, w, label='With Sentiment', color='steelblue', alpha=0.8)
    else:
        axes[0].bar(i - w/2, without_sent, w, color='lightcoral', alpha=0.8)
        axes[0].bar(i + w/2, with_sent, w, color='steelblue', alpha=0.8)

axes[0].set_xticks(x)
axes[0].set_xticklabels(crisis_pivot_sorted.index, rotation=30)
axes[0].set_ylabel('MAE (£/MWh)')
axes[0].set_title('Crisis Holdout - MAE Comparison')
axes[0].legend()

# R2 comparison
for i, model in enumerate(crisis_pivot_sorted.index):
    with_sent = crisis_pivot.loc[model, 'With Sentiment'] if 'With Sentiment' in crisis_pivot.columns else np.nan
    without_sent = crisis_pivot.loc[model, 'Without Sentiment'] if 'Without Sentiment' in crisis_pivot.columns else np.nan

r2_df = crisis_results.pivot_table(index='Model', columns='Config', values='R2', aggfunc='mean')
r2_with = r2_df['With Sentiment'] if 'With Sentiment' in r2_df.columns else None
r2_without = r2_df['Without Sentiment'] if 'Without Sentiment' in r2_df.columns else None

models_sorted = crisis_pivot_sorted.index
for i, model in enumerate(models_sorted):
    axes[1].bar(i - w/2, r2_without[model], w, color='lightcoral', alpha=0.8)
    axes[1].bar(i + w/2, r2_with[model], w, color='steelblue', alpha=0.8)

axes[1].axhline(0, color='gray', ls='--', linewidth=0.8)
axes[1].set_xticks(x)
axes[1].set_xticklabels(models_sorted, rotation=30)
axes[1].set_ylabel('R²')
axes[1].set_title('Crisis Holdout - R² Comparison')

plt.tight_layout()
plt.show()
```

Clear pattern here:
- Transformer + sentiment wins during crisis (R²=0.50, MAE=66.09)
- GatedTransformer is a disaster - too complex, overfits badly (R²=-0.35 with sentiment!)
- Sentiment helps the Transformer (MAE drops 4.1%) but hurts GatedTransformer
- ARIMA and XGBoost both have negative R² during crisis - they can't handle the regime shift

The GatedTransformer result is really interesting - adding more features actually made it WORSE. Probably because the gating mechanism wasn't regularised enough and it just memorised the noise.

```python
# Walk-forward - how do models perform as training data grows?
fig, ax = plt.subplots(figsize=(10, 5))

for model in wf_results['Model'].unique():
    subset = wf_results[wf_results['Model'] == model]
    subset = subset.groupby('Fold')['MAE'].mean()
    ax.plot(subset.index, subset.values, marker='o', label=model)

ax.set_xlabel('Fold')
ax.set_ylabel('MAE (£/MWh)')
ax.set_title('Walk-Forward: MAE by Fold (models improve with more data)')
ax.legend()
ax.set_xticks(range(1, 6))
plt.show()
```

This confirms what I suspected - deep learning models are data hungry.
- Fold 1: GatedTransformer and Transformer are terrible (R² like -10 and -3.5)
- By Fold 5: they converge and start working
- XGBoost is basically flat from Fold 1 - it doesn't need as much data
- ARIMA is consistently bad (as expected)

## 8. Sentiment Analysis: Does it actually help?

Let's look at the Diebold-Mariano test results if available, or at least compare the with/without sentiment differences.

```python
# Compare with vs without sentiment across models
print('=== CRISIS: Sentiment Impact ===')
for model in crisis_results['Model'].unique():
    subset = crisis_results[crisis_results['Model'] == model]
    if len(subset) >= 2:
        with_s = subset[subset['Config'] == 'With Sentiment']['MAE'].values[0]
        without_s = subset[subset['Config'] == 'Without Sentiment']['MAE'].values[0]
        diff = without_s - with_s
        pct = diff / without_s * 100
        print(f"{model:20s}: without={without_s:6.2f}  with={with_s:6.2f}  Δ={diff:+.2f} ({pct:+.1f}%)")

print()
print('=== WALK-FORWARD: Sentiment Impact (avg across folds) ===')
wf_avg = wf_results.groupby(['Model', 'Config'])['MAE'].mean().round(2).reset_index()
for model in wf_avg['Model'].unique():
    subset = wf_avg[wf_avg['Model'] == model]
    with_row = subset[subset['Config'] == 'With Sentiment']
    without_row = subset[subset['Config'] == 'Without Sentiment']
    if len(with_row) and len(without_row):
        with_s = with_row['MAE'].values[0]
        without_s = without_row['MAE'].values[0]
        diff = without_s - with_s
        pct = diff / without_s * 100
        print(f"{model:20s}: without={without_s:6.2f}  with={with_s:6.2f}  Δ={diff:+.2f} ({pct:+.1f}%)")
```

Key takeaways:

**Crisis period:**
- Transformer: sentiment helps (MAE -4.1%)
- XGBoost: marginal improvement (-2.7%)
- LSTM: basically flat
- GatedTransformer: sentiment actively hurts (-28.4% worse!)

**Normal conditions (walk-forward):**
- Sentiment is neutral to slightly detrimental for most models
- Transformer with sentiment actually does worse (-10% MAE in normal conditions)

This aligns with the hypothesis: sentiment only adds value during market stress when historical patterns break down. During normal conditions, price history + grid data is sufficient.

## 9. Actual vs Predicted (Crisis Period)

Let's see how the best model (Transformer + sentiment) tracks actual prices during the crisis.

```python
preds = pd.read_csv(os.path.join(PROCESSED, 'forecast_predictions_crisis.csv'), parse_dates=['datetime_local'])

# Subsample for visibility - first 7 days
week1 = preds.iloc[:336].copy()  # 336 = 7 days * 48 half-hours

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(week1['datetime_local'], week1['systemSellPrice'], label='Actual', color='black', linewidth=1.2)
ax.plot(week1['datetime_local'], week1['pred_Transformer_With_Sentiment'], label='Transformer + Sentiment', 
        color='steelblue', linewidth=0.8, alpha=0.8)
ax.plot(week1['datetime_local'], week1['pred_XGBoost_With_Sentiment'], label='XGBoost + Sentiment', 
        color='orange', linewidth=0.8, alpha=0.8)
ax.plot(week1['datetime_local'], week1['pred_ARIMA'], label='ARIMA', 
        linewidth=0.8, alpha=0.6)

ax.set_title('Actual vs Predicted - First Week of Crisis (Feb 2022)')
ax.set_ylabel('£/MWh')
ax.legend()
plt.show()
```

```python
# Mean absolute error over the crisis (rolling 7-day window)
preds['transformer_error'] = abs(preds['systemSellPrice'] - preds['pred_Transformer_With_Sentiment'])
preds['xgb_error'] = abs(preds['systemSellPrice'] - preds['pred_XGBoost_With_Sentiment'])
preds['arima_error'] = abs(preds['systemSellPrice'] - preds['pred_ARIMA'])

window = 7 * 48  # 7 days of half-hourly
preds['transformer_mae_7d'] = preds['transformer_error'].rolling(window).mean()
preds['xgb_mae_7d'] = preds['xgb_error'].rolling(window).mean()
preds['arima_mae_7d'] = preds['arima_error'].rolling(window).mean()

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(preds['datetime_local'], preds['transformer_mae_7d'], label='Transformer + Sentiment', color='steelblue')
ax.plot(preds['datetime_local'], preds['xgb_mae_7d'], label='XGBoost + Sentiment', color='orange')
ax.plot(preds['datetime_local'], preds['arima_mae_7d'], label='ARIMA', color='gray')
ax.set_title('Rolling 7-Day MAE During Crisis (Transformer holds up best)')
ax.set_ylabel('MAE (£/MWh)')
ax.legend()
plt.show()
```

The Transformer tracks the price movements much better than XGBoost or ARIMA during the crisis.
ARIMA is basically useless (it just predicts the mean). XGBoost captures some of the variation but has larger spikes of error.

## 10. Summary of Key Findings

Key numbers to report:
| Finding | Value |
|---------|-------|
| Best crisis model | Transformer + Sentiment (R²=0.50, MAE=66.09) |
| Sentiment improvement (crisis) | -4.1% MAE for Transformer |
| Best normal conditions | XGBoost (R²=0.66, MAE=17.82) |

| NLP Sentiment contribution | 6.9% (modest but non-trivial) |
| Biggest feature overall | Price lag_1 (24.1% of importance) |

The bottom line: multimodal fusion helps during regime shifts. During normal periods, keep it simple.
